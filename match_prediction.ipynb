```python
from model import Model
from datasets import load_dataset
from langchain_core.prompts import PromptTemplate
```

```python
dataset = load_dataset('csv', data_files='data/matches_processed.csv', split={"train": "train[:99%]", "test": "train[:1%]"})

def format_prompt(input: str):
    prompt_template = PromptTemplate.from_template(open("./prompt_templates/win_prediction.md", "r").read())
    return {"prompt": prompt_template.invoke({"input": input["Player"]}).to_string()}

dataset = dataset.map(format_prompt)   

dataset = dataset.remove_columns(["MatchId", "Player"])
```
